```python
import pandas as pd
#loading dataset
df = pd.read_csv("C:\\Users\\kusha\\OneDrive\\Desktop\\DA\\My projects\\Marketing Campaign Analysis\\raw dataset\\global_ads_performance_dataset.csv")

#Table overview
#returning dimension of dataframe
print("Rows:", df.shape[0])  
print("Columns:", df.shape[1]) 
df.head()
```

```python
#Column Information 
df.info()
```

```python
#Checkig unique values in categorical columns
categorical_columns = [

    'platform', 
    'campaign_type', 
    'industry', 
    'country'
]

for col in categorical_columns: 
    print(f"\n{col}:")
    print(df[col].unique())
```

```python
#Statistical overview of numeric columns 
df.describe()
```

```python
#Checking date raneg and number of unique dates 
print("Minimum date:", df['date'].min())
print("Maximum date:", df['date'].max())
print("Unique date:", df['date'].nunique())
```

```python
# Checking number of records per date
df['date'].value_counts().describe()
```

```python
#checking for completely duplicate rows
print("duplicate rows:", df.duplicated().sum())
```

```python
#checking whether date + platform + campaign type + industry +country uniquely identifies each record
key_columns = [
    'date',
    'platform',
    'campaign_type',
    'industry',
    'country'
]

print("Duplicate combinations:", df.duplicated(subset=key_columns).sum())
```

```python
#inspecting duplicate combinations
duplicates = df[df.duplicated(
    subset = key_columns, 
    keep = False
)].sort_values(key_columns)

duplicates
```

```python
# Checking frequency of repeated dimension combinations

df.groupby(key_columns).size().value_counts().sort_index()
```

```python
# Checking uniqueness without date

non_date_key = [
    'platform',
    'campaign_type',
    'industry',
    'country'
]

print(
    "Duplicate combinations:",
    df.duplicated(subset=non_date_key).sum()
)
```

```python
# Checking whether adding ad_spend makes the combination unique
extended_key = [
    'date',
    'platform',
    'campaign_type',
    'industry',
    'country',
    'ad_spend'
]

print(
    "Duplicate combinations:",
    df.duplicated(subset=extended_key).sum()
)
```

```python
# Comparing ad spend within repeated date-dimension combinations

duplicates.groupby(key_columns)['ad_spend'].nunique()
```
